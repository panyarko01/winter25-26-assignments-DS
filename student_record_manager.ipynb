```python
class student:
    def __init__(self, name, student_id, courses_and_grades):
        self.name = name
        self.student_id = student_id
        self.courses_and_grades = courses_and_grades

    def display(self):
        print(f"Name: {self.name}, Student_ID: {self.student_id}, Courses and Grades: {self.courses_and_grades}")

    def get_average_grade(self):
        total = 0.
        for grade in self.courses_and_grades.values():
            total += grade
        average = total / len(self.courses_and_grades)
        return average




Max = student("Max", "12345", {"Math":5, "Science": 2})
Max.display()
print("Average grade is",Max.get_average_grade())
```
